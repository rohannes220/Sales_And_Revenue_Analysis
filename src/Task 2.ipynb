```python
import pandas as pd

orders = pd.read_csv("../data/orders.csv")
shipments = pd.read_csv("../data/shipments.csv")
display(orders.head())
display(shipments.head())
```

```python
print("Orders order_id unique:", orders["order_id"].is_unique)
print("Shipments order_id unique:", shipments["order_id"].is_unique)
```

- `order_id` is unique in the orders dataset.
- `order_id` is not unique in the shipments dataset.
- Therefore, the join has a one-to-many relationship: one order can have multiple shipments.
