In this section, I will examine the three datasets in their original raw form.

```python
revenue = pd.read_csv("../data/company_quarterly_revenue_raw.csv")
orders = pd.read_csv("../data/orders.csv")
shipments = pd.read_csv("../data/shipments.csv")
```

### Tidiness Assessment

#### Revenue Dataset
The revenue dataset has 5 rows and 13 columns. It is **not tidy**.

- **Rule 1: Each variable should have its own column — Violated.** Quarter is a variable, but its values (2023-Q1, 2023-Q2, etc.) are stored as separate column names instead of in a single `quarter` column.
- **Rule 2: Each observation should have its own row — Violated.** A company-quarter should represent one observation, but multiple quarterly observations are stored across a single company row.
- **Rule 3: Each value should have its own cell — Satisfied.** Each individual revenue value is stored in its own cell.

Therefore, the revenue dataset is not tidy and will need to be reshaped from wide to long format.

#### Orders Dataset
The orders dataset has 20 rows and 3 columns. It is **tidy**.

- **Rule 1: Satisfied.** The variables `order_id`, `customer_id`, and `amount` each have their own column.
- **Rule 2: Satisfied.** Each row represents one order.
- **Rule 3: Satisfied.** Each value is stored in its own cell.

Therefore, the orders dataset follows the three rules of tidy data.

#### Shipments Dataset
The shipments dataset has 35 rows and 2 columns. It is **tidy**.

- **Rule 1: Satisfied.** The variables `order_id` and `tracking` each have their own column.
- **Rule 2: Satisfied.** Each row represents one shipment.
- **Rule 3: Satisfied.** Each value is stored in its own cell.

Therefore, the shipments dataset follows the three rules of tidy data.

### Exploratory Visualization of Raw Data

The following visualization uses the raw revenue dataset before any reshaping or tidying. It compares company revenue for the first quarter of 2023.

```python
import matplotlib.pyplot as plt
plt.bar(revenue["company"], revenue["2023-Q1"])

plt.xlabel("Company")
plt.ylabel("Revenue")
plt.title("Company Revenue in 2023-Q1")

plt.show()
```
