# Task 4: Analysis-Ready Table

- Load the revenue, orders, and shipments datasets.
- Reshape the revenue data into its tidy format.
- Recreate the correctly joined orders and shipments table.
- Combine the tidy revenue data and joined order data into one analysis-ready table.
- Make sure the final table has sensible column names, one row per observation, and no leftover index artifacts.

```python
import pandas as pd

revenue = pd.read_csv("../data/company_quarterly_revenue_raw.csv")
orders = pd.read_csv("../data/orders.csv")
shipments = pd.read_csv("../data/shipments.csv")
```

```python
revenue_tidy = revenue.melt(
    id_vars="company",
    var_name="quarter",
    value_name="revenue"
)
```

```python
joined_orders = orders.merge(
    shipments,
    on="order_id",
    how="left",
    validate="one_to_many"
)
```

```python
display(revenue_tidy.head())
display(joined_orders.head())

print(revenue_tidy.columns)
print(joined_orders.columns)
```

```python
final_table = joined_orders.merge(
    revenue_tidy,
    on=["company", "quarter"],
    how="left"
)

final_table = final_table.reset_index(drop=True)

display(final_table.head())
print("Joined orders rows:", len(joined_orders))
print("Final table rows:", len(final_table))
```
