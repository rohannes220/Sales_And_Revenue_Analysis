# Task 3: Join Orders and Shipments

- Join the orders and shipments datasets using `order_id`.
- Calculate the mean order amount per customer using the joined table.
- Calculate the mean order amount per customer using the original orders table.
- Compare the results and determine which calculation is correct.

```python
import pandas as pd

orders = pd.read_csv("../data/orders.csv")
shipments = pd.read_csv("../data/shipments.csv")
```

```python
display(orders.head())
display(shipments.head())
```

```python
joined = orders.merge(
    shipments,
    on="order_id",
    how="left",
    validate="one_to_many"
)
```

```python
joined.head(10)
```

```python
#average on the joined table
joined_average = joined.groupby('customer_id')['amount'].mean()
#average on the non joined table
regular_average = orders.groupby("customer_id")["amount"].mean()
display(joined_average)
display(regular_average)
```

### Task 3 Conclusion

- The mean amount per customer differs between the joined table and the original orders table for some customers.
- This happens because the join is one-to-many. An order can have multiple shipments, so some orders appear multiple times in the joined table.
- When the mean is calculated directly from the joined table, the amount for orders with multiple shipments is counted multiple times.
- The mean calculated from the orders table is correct because each order appears only once, so every order amount is counted once.
