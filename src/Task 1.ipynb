# Task 1: Tidy the Revenue Table

In this task, I will examine whether the revenue dataset is stored in long or wide format. If it is wide, I will reshape it so that each row represents one company-quarter observation.

The raw revenue dataset violates the first rule of tidy data because the `quarter` variable does not have its own column. Instead, quarter values such as `2023-Q1`, `2023-Q2`, and `2023-Q3` are stored as column names. It also violates the second rule because multiple company-quarter observations are stored in a single row.

I will reshape the dataset so that the final columns are `company`, `quarter`, and `revenue`.

```python
import pandas as pd

revenue = pd.read_csv("../data/company_quarterly_revenue_raw.csv")
print(revenue.shape)
revenue.head()
```

# Task 1: Revenue Table

- The revenue dataset is in wide format.
- It violates Rule 1 of tidy data: each variable should have its own column.
- Quarter is a variable, but values such as 2023-Q1, 2023-Q2, etc. are stored as column names instead of in one quarter column.
- It also violates Rule 2: each observation should have its own row, because multiple company-quarter observations are stored in one row.
- I will reshape the data into long format so that each row represents one (company, quarter) observation.

```python
melted_df = revenue.melt(id_vars='company',var_name = 'quarter', value_name = 'revenue')
```

```python
melted_df.head(4)
```

- The revenue table has been reshaped from wide to long format.
- Each row now represents one (company, quarter) observation.
- The variables company, quarter, and revenue each have their own column.
- The resulting revenue table is tidy.
